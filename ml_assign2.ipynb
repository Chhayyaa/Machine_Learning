```python
selected_cols=["MaritalStatus",
               "Gender",
               "YearlyIncome",
               "TotalChildren",
               "Education",
               "Occupation",
               "HomeOwnerFlag",
               "NumberCarsOwned",
               "CountryRegionName"]
df_selected =df[selected_cols]
```

```python
#numerical columns
num_cols = df_selected.select_dtypes(include=['int64','float64']).columns
df_selected[num_cols] = df_selected[num_cols].fillna(df_selected[num_cols].median())
#categorical columns
cat_cols = df_selected.select_dtypes(include=['object']).columns
df_selected[cat_cols] = df_selected[cat_cols].fillna(df_selected[cat_cols].mode().iloc[0])
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df_selected[['YearlyIncome']] = scaler.fit_transform(
    df_selected[['YearlyIncome']]
)
```

```python
df_selected['Income Bin'] = pd.qcut(
    df_selected['YearlyIncome'],
    q=4,
    labels=['Low','Medium','High','Very High']
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_selected[['YearlyIncome']] = scaler.fit_transform(
    df_selected[['YearlyIncome',]]
)
```

```python
df_final = pd.get_dummies(
    df_selected,
    columns=[
        'Gender',
        'MaritalStatus',
        'HomeOwnerFlag'
    ],
    drop_first=True
)
```
